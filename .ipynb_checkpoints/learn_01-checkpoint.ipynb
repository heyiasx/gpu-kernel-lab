```python
# 原样拷贝
import torch
import triton
import triton.language as tl

@triton.jit
def demo(x_ptr, out_ptr):  # 指向第 0 个元素的地址
    offs = tl.arange(0, 8)  # 索引向量
    mask = offs < 5
    x = tl.load(x_ptr + offs, mask=mask, other=0.0)  # 从 x_ptr 指向的内存读取 offs 指定的位置。
    tl.store(out_ptr + offs, x)  # 把 x 写回 out_ptr + offs。

x = torch.arange(8, dtype=torch.float32, device='cuda')
out = torch.empty_like(x)  # 分配一个同样形状的 GPU tensor，未初始化，内容是随机的。

demo[(1,)](x, out)  # (1,) 是 grid，表示只启动 1 个 program，网格
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print("out =", out)
```

```python
# 4*3的二维张量，在内存中也是连续存放的
x2 = torch.ones(4,3,dtype=torch.float32,device="cuda")
demo[(1,1,1)](x2,out)
torch.cuda.synchronize()
print(out)
```

```python
# 两个向量求和
import triton
import torch
import triton.language as tl

@triton.jit
def vector_add_kernel(x_ptr,y_ptr,output_ptr,n_elements,BLOCK_SIZE:tl.constexpr):
    pid = tl.program_id(axis=0)
    block_start = pid * BLOCK_SIZE
    offsets = block_start + tl.arange(0,BLOCK_SIZE)
    mask = offsets < n_elements

    x = tl.load(x_ptr+offsets,mask=mask)
    y = tl.load(y_ptr+offsets,mask=mask)
    output = x + y
    tl.store(output_ptr+offsets,output,mask=mask)

def vector_add(x:torch.Tensor,y:torch.Tensor) -> torch.Tensor:
    output = torch.empty_like(x)
    n_elements = output.numel()
    grid = lambda meta:(triton.cdiv(n_elements,meta["BLOCK_SIZE"]),)  # 这里要加一个逗号，因为返回的值一定要是一个元组，不加逗号返回的就是一个整数
    vector_add_kernel[grid](x,y,output,n_elements,BLOCK_SIZE=1024)
    return output

x = torch.randn(4096,device="cuda")
y = torch.randn(4096,device="cuda")
output = vector_add(x,y)
print(x)
print(y)
print(output)
```

```python
# triton的数据类型与操作：tl.cast(变量,tl.float16)
import triton
import triton.language as tl

@triton.jit
def cast_example(x_ptr,y_ptr,n):
    pid = tl.program_id(axis=0)
    offsets = pid*128 + tl.arange(0,128)
    mask = offsets < n

    # 从FP32读到，读取加载数据
    x = tl.load(x_ptr+offsets,mask=mask)
    tl.static_print("x dtype = ",x.dtype)
    
    # 转成FP16计算，运算
    x_fp16 = tl.cast(x,tl.float16)  # FP32转FP16
    y_fp16 = x_fp16*2.0
    
    # 编译期打印 dtype（终端在编译时看到）
    tl.static_print("y_fp16 dtype = ", y_fp16.dtype)

    # 转回FP32写出去，写回
    y = tl.cast(y_fp16,tl.float32)  # FP16转回FP32
    tl.static_print("y dtype = ",y.dtype)
    tl.store(y_ptr+offsets,y,mask=mask)  # 把y写回这个地址

x = torch.randn(10,device="cuda")
y = torch.randn(10,device="cuda")
print(x.dtype)
cast_example[(1,)](x,y,7)
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print(y.dtype)
```

```python
# triton的算数操作
@triton.jit
def arith_example(a_ptr,b_ptr,c_ptr,n):
    pid = tl.program_id(axis=0)
    offsets = pid * 128 + tl.arange(0,128)
    mask = offsets < n

    a = tl.load(a_ptr+offsets,mask=mask)
    b = tl.load(b_ptr+offsets,mask=mask)

    # 基本运算
    add = a + b
    sub = a - b
    mul = a * b
    div = a / b  # 注意整数除法会转换成浮点
    

    # 取模
    mod = a % b

    tl.store(c_ptr+offsets,div,mask=mask) # 把add的结果存入c_ptr指针指向的位置

a = torch.randn(4,device="cuda")
b = torch.randn(4,device="cuda")
c = torch.empty_like(x)
arith_example[(1,)](a,b,c,3)
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print(a)
print(b)
print(c)

    
```

```python
# triton中使用数学函数
@triton.jit
def math_demo(x_ptr,y_ptr,n):  # 传入x变量和y变量第一个元素的地址
    pid = tl.program_id(axis=0)
    offsets = pid*128 + tl.arange(0,128)
    mask = offsets < n

    x = tl.load(x_ptr+offsets,mask=mask)  # tl.load(地址)

    # 常用的数学操作
    y1 = tl.exp(x)  # y=e^x
    y2 = tl.log(x)  # ln(x)
    y3 = tl.sqrt(x)  # sqrt(x)
    y4 = tl.rsqrt(x)  # 1/sqrt(x) 这在归一化LayerNorm中很常用，性能比1.0/tl.sqrt(x)快
    y5 = tl.abs(x)   # |x|

    tl.store(y_ptr+offsets,y5,mask=mask)  # tl.store(地址，数据)

x = torch.tensor([16.0],device="cuda")  # 一维张量，shape(1,)
print(x)
y = torch.ones_like(x)   # 创建一个和 x 形状、数据类型、设备都相同的新 tensor，但不初始化它的内容。
print(y)
n = x.numel()  # n为x张量的元素个数
math_demo[(1,)](x,y,n)
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print(y)
```

```python
# 比较操作
@triton.jit
def compare_demo(a_ptr,b_ptr,mask_ptr,n):
    pid = tl.program_id(axis=0)
    offsets = pid * 128 + tl.arange(0,128)
    mask = offsets < n 

    a = tl.load(a_ptr + offsets,mask=mask)
    b = tl.load(b_ptr + offsets,mask=mask)

    # 比较操作
    eq = a == b  # 等于
    ne = a != b  # 不等于
    lt = a < b # 小于
    gt = a > b # 大于
    le = a <= b # 小于等于
    ge = a >= b # 大于等于

    tl.store(mask_ptr+offsets,ne,mask=mask)

a = torch.tensor(16,device="cuda")
b = torch.tensor(12,device="cuda")
mask_ptr = torch.ones_like(x)
n = x.numel()
compare_demo[(1,)](a,b,mask_ptr,n)
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print(mask_ptr)  # 0为False,1为True
```

```python
# 用布尔条件做逻辑运算
import torch
import triton
import triton.language as tl

@triton.jit
def logic_demo(x_ptr, y_ptr, z_ptr, n, BLOCK: tl.constexpr):
    pid = tl.program_id(axis=0)
    offs = pid * BLOCK + tl.arange(0, BLOCK)
    mask = offs < n

    x = tl.load(x_ptr + offs, mask=mask, other=0.0)
    y = tl.load(y_ptr + offs, mask=mask, other=0.0)

    # 先转成布尔
    x_bool = x > 0
    y_bool = y > 0

    and_op = x_bool & y_bool     # 布尔与
    or_op  = x_bool | y_bool     # 布尔或
    not_op = ~x_bool             # 布尔非

    # 用条件选择输出：x>0 选 x，否则选 y
    result = tl.where(x > 0, x, y)

    tl.store(z_ptr + offs, not_op, mask=mask)

x = torch.tensor([1.0], device="cuda")
y = torch.tensor([0.0], device="cuda")
z = torch.zeros_like(x)
n = x.numel()

logic_demo[(1,)](x, y, z, n, BLOCK=128)
torch.cuda.synchronize()  # 等待 GPU 执行完成，保证后面的 print 看到的是最终结果。
print(z)   # tensor([1.], device='cuda:0')  —— x>0，所以选 x
```

```python
#用整数演示真正的按位运算
@triton.jit
def bitwise_demo(x_ptr, y_ptr, z_ptr, n, BLOCK: tl.constexpr):
    pid = tl.program_id(axis=0)
    offs = pid * BLOCK + tl.arange(0, BLOCK)
    mask = offs < n

    x = tl.load(x_ptr + offs, mask=mask, other=0)   # int32
    y = tl.load(y_ptr + offs, mask=mask, other=0)   # int32

    and_op = x & y
    or_op  = x | y
    not_op = ~x

    tl.store(z_ptr + offs, and_op, mask=mask)

x = torch.tensor([0b1100], dtype=torch.int32, device="cuda")
y = torch.tensor([0b1010], dtype=torch.int32, device="cuda")
z = torch.empty_like(x)
n = x.numel()

bitwise_demo[(1,)](x, y, z, n, BLOCK=128)
torch.cuda.synchronize()
print(z)   # tensor([8], dtype=torch.int32)  —— 0b1100 & 0b1010 = 0b1000 = 8
```
